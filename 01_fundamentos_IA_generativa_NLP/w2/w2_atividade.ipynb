# Importar as bibliotecas e carregar a LLM

````python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "mistralai/Mistral-7B-Instruct-v0.3"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
````

# Entendendo o Tokenizador

````python
prompt = "Diga as qualiades do LLM mistral."
inputs = tokenizer(prompt)
# imprime os tokens de entrada
print(inputs)
# converte os tokens de entrada em um tensor
inputs_tensor = tokenizer(prompt, return_tensors="pt")
# imprime o tensor de entrada
print(inputs_tensor)
````

# Criando um Pipeline

````python
from transformers import pipeline
# Create a pipeline
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    return_full_text=False,
    max_new_tokens=500,
    do_sample=False
)
````

````python
# Werner Denzin | Adicionado trecho de código para evitar os *warnings* acima

from transformers import pipeline, GenerationConfig

model.generation_config = GenerationConfig(
    max_new_tokens=500,
    do_sample=False
)

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    return_full_text=False
)
````

# Gerando uma resposta

````python
# Generate output
output = generator(prompt)
print(output[0]["generated_text"])
````

**5. Entenda o funcionamento do LM_Head:**

````python
prompt2 = "The capital of USA is"                               # Prompt do usuário
input_ids = tokenizer(prompt2, return_tensors="pt").input_ids   # Convert prompt em token IDs
input_ids = input_ids.to('cuda')                                # Configuração para uso de GPU (ganho de performance)
model_output = model.model(input_ids)                           # Seria apenas a extração de (model) de sua arquitetura
lm_head_output = model.lm_head(model_output[0])                 # Seria apenas a extração da (lm head) última camada da arquitetura
tokens_id = lm_head_output[0,-1].argmax(dim=-1)                 # argmax() retorna o último token e portanto, com maior score = próximo token!
resposta = tokenizer.decode(tokens_id)                          # Converte token ID para a respectiva palavra
print(resposta)
````

````python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "microsoft/Phi-3-mini-4k-instruct"
tokenizer  = AutoTokenizer.from_pretrained(model_name)
model      = AutoModelForCausalLM.from_pretrained(model_name)
````

````python
from transformers import pipeline

prompt = "Diga as qualiades do LLM mistral."

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

output = generator(
    prompt,
    max_new_tokens=1000,
    temperature=0.2,
    do_sample=True
)

print(output[0]["generated_text"])
````
